```python
# @title Research Paper Data Analysis (Corrected Font Link)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import glob
import os
import matplotlib.font_manager as fm
import urllib.request

# ==========================================
# 1. FONT SETUP (Times New Roman)
# ==========================================
# We download the font locally to ensure it works in Google Colab
font_filename = 'Times_New_Roman.ttf'
font_url = "https://github.com/justrajdeep/fonts/raw/master/Times%20New%20Roman.ttf"

if not os.path.exists(font_filename):
    print(f"⬇️ Downloading font from {font_url}...")
    try:
        urllib.request.urlretrieve(font_url, font_filename)
        print("✅ Font downloaded successfully.")
    except Exception as e:
        print(f"⚠️ Font download failed: {e}")
        print("   Using default serif font instead.")

# Register the font with Matplotlib
if os.path.exists(font_filename):
    fm.fontManager.addfont(font_filename)
    plt.rcParams['font.family'] = 'serif'
    plt.rcParams['font.serif'] = ['Times New Roman']
else:
    plt.rcParams['font.family'] = 'serif'

# Global Plot Styling (Research Paper Standard)
plt.rcParams.update({
    'font.size': 12,
    'axes.labelsize': 12,
    'axes.titlesize': 14,
    'xtick.labelsize': 11,
    'ytick.labelsize': 11,
    'legend.fontsize': 11,
    'figure.titlesize': 16,
    'axes.grid': True,
    'grid.alpha': 0.3,
    'grid.linestyle': '--'
})

# ==========================================
# 2. ROBUST DATA LOADING (Auto-Detect File)
# ==========================================
# Finds the first .csv or .xlsx file in the folder (excluding system files)
possible_files = glob.glob("*.xlsx") + glob.glob("*.csv")
data_files = [f for f in possible_files if "sample_data" not in f and "Times" not in f]

if not data_files:
    print("❌ Error: No dataset found. Please upload your Excel or CSV file.")
else:
    file_path = data_files[0]
    print(f"✅ Dataset detected: {file_path}")

    try:
        # Load based on extension
        if file_path.endswith('.xlsx') or file_path.endswith('.xls'):
            df = pd.read_excel(file_path)
        else:
            df = pd.read_csv(file_path)

        # Data Cleaning
        df = df.dropna(axis=1, how='all').dropna(how='all')
        df.columns = df.columns.str.strip()
        print("✅ Data loaded and cleaned.")

    except Exception as e:
        print(f"❌ Error reading file: {e}")

# ==========================================
# 3. DEFINE FORMAL LABELS
# ==========================================
labels = {
    'Scaled stand off distance': r'Scaled Standoff Distance $Z$ ($\mathrm{m/kg}^{1/3}$)',
    'Axial load ratio': r'Axial Load Ratio $n$ (-)',
    'Confinement': r'Confinement Factor $\xi$ (-)',
    'Loading interaction': r'Loading Interaction (-)',
    'Aspect ratio': r'Aspect Ratio (-)',
    'Load confinement interaction': r'Load-Confinement Interaction (-)',
    'Peak deformation': r'Peak Deformation $\delta$ (mm)'
}

# Map columns
plot_df = pd.DataFrame()
for col in df.columns:
    if col in labels:
        plot_df[labels[col]] = df[col]
    else:
        plot_df[col] = df[col]

target_col = labels.get('Peak deformation', 'Peak deformation')
feature_cols = [col for col in plot_df.columns if col != target_col]

# ==========================================
# 4. PLOTS
# ==========================================

# --- Plot 1: Correlation Matrix ---
if len(plot_df) > 0:
    plt.figure(figsize=(10, 8))
    numeric_df = plot_df.select_dtypes(include=[np.number])
    corr_matrix = numeric_df.corr(method='pearson')

    sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm',
                linewidths=0.5, square=True, cbar_kws={"shrink": .8})
    plt.title('Pearson Correlation Matrix', fontsize=14, pad=20)
    plt.tight_layout()
    plt.show()

# --- Plot 2: Box Plots ---
if len(plot_df) > 0:
    plt.figure(figsize=(16, 10))
    for i, col in enumerate(feature_cols):
        plt.subplot(2, 3, i + 1)
        sns.boxplot(y=plot_df[col], color='skyblue', width=0.5, linewidth=1.5)

        clean_title = col.split('$')[0].strip()
        plt.title(f'Distribution of {clean_title}', fontsize=13)
        plt.ylabel(col, fontweight='bold')
        plt.grid(axis='y', linestyle='--', alpha=0.7)

    plt.suptitle('Statistical Distribution of Input Variables', fontsize=16, y=1.02)
    plt.tight_layout()
    plt.show()

# --- Plot 3: Scatter Plots (No Regression Line) ---
if len(plot_df) > 0:
    plt.figure(figsize=(16, 10))
    for i, col in enumerate(feature_cols):
        plt.subplot(2, 3, i + 1)
        sns.scatterplot(x=plot_df[col], y=plot_df[target_col],
                        alpha=0.7, s=80, edgecolor='black', color='coral')

        clean_title = col.split('$')[0].strip()
        plt.title(f'{clean_title} vs. Peak Deformation', fontsize=13)
        plt.xlabel(col, fontweight='bold')
        plt.ylabel(target_col, fontweight='bold')
        plt.grid(True, linestyle='--', alpha=0.5)

    plt.suptitle('Scatter Plots: Input Variables vs. Peak Deformation', fontsize=16, y=1.02)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import glob
import os

# ==========================================
# 1. SETUP PROFESSIONAL STYLING
# ==========================================
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman', 'DejaVu Serif']
plt.rcParams['font.size'] = 12

# ==========================================
# 2. ROBUST DATA LOADING
# ==========================================
possible_files = glob.glob("*.xlsx") + glob.glob("*.csv")
data_files = [f for f in possible_files if "sample_data" not in f]

if not data_files:
    print("❌ Error: No dataset found. Please upload 'processed_data2.xlsx'.")
else:
    file_path = data_files[0]
    print(f"✅ Dataset detected: {file_path}")

    try:
        if file_path.endswith('.xlsx'):
            df = pd.read_excel(file_path)
        else:
            df = pd.read_csv(file_path)

        # Clean column names
        df.columns = df.columns.str.strip()
        print(f"✅ Columns found: {df.columns.tolist()}")

        # ==========================================
        # 3. SELECT ALL NUMERIC VARIABLES
        # ==========================================
        # This ensures WE DO NOT MISS ANY PARAMETER in the Excel file
        table_df = df.select_dtypes(include=[np.number])

        # Define Professional Labels Mapping
        labels = {
            'dimensions': 'Dimension D (mm)',
            'span length of specimen': 'Span Length L (mm)',
            'grade of concrete(Mpa)': 'Concrete Grade f_c (MPa)',
            'yield strength of the steel (Mpa)': 'Steel Yield f_y (MPa)',
            'Rx2': 'Rx2 (-)',
            'scaled stand off distance': 'Standoff Z (m/kg⅓)',
            'Scaled stand off distance': 'Standoff Z (m/kg⅓)', # Variance case
            'axial load ratio': 'Axial Load Ratio n (-)',
            'Axial load ratio': 'Axial Load Ratio n (-)',     # Variance case
            'confinement': 'Confinement ξ (-)',
            'Confinement': 'Confinement ξ (-)',             # Variance case
            'peak deformation (mm)': 'Peak Def. δ (mm)',
            'Peak deformation': 'Peak Def. δ (mm)',         # Variance case
            'shape': 'Shape Factor (-)',
            'aspect_ratio': 'Aspect Ratio (-)',
            'Aspect ratio': 'Aspect Ratio (-)',             # Variance case
            'load_confinement_interaction': 'Load-Conf. Interaction (-)',
            'Load confinement interaction': 'Load-Conf. Interaction (-)', # Variance case
            'loading_interaction': 'Loading Interaction (-)',
            'Loading interaction': 'Loading Interaction (-)' # Variance case
        }

        # Apply renaming (Only updates names that match, keeps others as is)
        table_df = table_df.rename(columns=labels)

        # ==========================================
        # 4. CALCULATE STATISTICS (Min, Max, Mean, Std, Skew, Kurt)
        # ==========================================
        stats = table_df.describe().T
        stats['Skewness'] = table_df.skew()
        stats['Kurtosis'] = table_df.kurt()

        # Select and Reorder Columns for the Paper
        final_cols = ['min', '25%', '50%', '75%', 'max', 'mean', 'std', 'Skewness', 'Kurtosis']
        final_table = stats[final_cols]

        # Rename Headers for Professional Look
        header_renames = {
            'min': 'Min',
            '25%': 'Q1 (25%)',
            '50%': 'Median',
            '75%': 'Q3 (75%)',
            'max': 'Max',
            'mean': 'Mean',
            'std': 'St. Dev',
            'Skewness': 'Skewness',
            'Kurtosis': 'Kurtosis'
        }
        final_table = final_table.rename(columns=header_renames)

        # Round to 2 decimal places
        plot_data = final_table.round(2).reset_index()
        plot_data.rename(columns={'index': 'Variable'}, inplace=True)

        # ==========================================
        # 5. GENERATE PROFESSIONAL TABLE IMAGE
        # ==========================================
        def render_mpl_table(data, col_width=2.0, row_height=0.625, font_size=12,
                             header_color='#40466e', row_colors=['#f1f1f2', 'w'], edge_color='w',
                             bbox=[0, 0, 1, 1], header_columns=0,
                             ax=None, **kwargs):
            if ax is None:
                # Dynamic height based on number of variables
                size = (np.array(data.shape[::-1]) + np.array([0, 1])) * np.array([col_width, row_height])
                fig, ax = plt.subplots(figsize=size)
                ax.axis('off')

            mpl_table = ax.table(cellText=data.values, bbox=bbox, colLabels=data.columns, **kwargs)
            mpl_table.auto_set_font_size(False)
            mpl_table.set_fontsize(font_size)

            for k, cell in mpl_table._cells.items():
                cell.set_edgecolor(edge_color)
                if k[0] == 0 or k[1] < header_columns:
                    cell.set_text_props(weight='bold', color='w')
                    cell.set_facecolor(header_color)
                else:
                    cell.set_facecolor(row_colors[k[0] % len(row_colors)])
            return ax

        # Render Table
        render_mpl_table(plot_data, header_columns=0, col_width=2.2)
        plt.title("Table 1: Comprehensive Statistical Description (All Variables)",
                  fontsize=16, y=1.02, fontweight='bold')
        plt.show()

        # ==========================================
        # 6. PRINT EXPLANATION
        # ==========================================
        print("\n" + "="*60)
        print("EXPLANATION OF STATISTICAL PARAMETERS")
        print("="*60)
        print("""
        1. Min (Minimum): The smallest value observed in the dataset.
        2. Q1 (25th Percentile): 25% of the data falls below this value.
        3. Median (50th Percentile): The robust central value of the dataset.
        4. Q3 (75th Percentile): 75% of the data falls below this value.
        5. Max (Maximum): The largest value observed.
        6. Mean: The arithmetic average.
        7. St. Dev: Variation/Dispersion. High values = spread out data.
        8. Skewness: Asymmetry. 0 = Normal, >0 = Right-skewed, <0 = Left-skewed.
        9. Kurtosis: Tailedness. 0 = Normal, >0 = Heavy tails (Outliers), <0 = Light tails.
        """)

    except Exception as e:
        print(f"❌ Error processing file: {e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import glob
import math

# Set professional font settings
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman', 'DejaVu Serif']
plt.rcParams['font.size'] = 12

# ==========================================
# 1. ROBUST DATA LOADING
# ==========================================
possible_files = glob.glob("*.xlsx") + glob.glob("*.csv")
data_files = [f for f in possible_files if "sample_data" not in f and "Times" not in f]

if not data_files:
    print("❌ Error: No dataset found. Please upload your Excel or CSV file.")
else:
    file_path = data_files[0]
    print(f"✅ Dataset detected: {file_path}")

    try:
        if file_path.endswith('.xlsx') or file_path.endswith('.xls'):
            df = pd.read_excel(file_path)
        else:
            df = pd.read_csv(file_path)

        # Clean column names
        df.columns = df.columns.str.strip()

        # ==========================================
        # 2. AUTOMATICALLY SELECT ALL NUMERIC VARIABLES
        # ==========================================
        vars_to_plot = df.select_dtypes(include=[np.number]).columns.tolist()
        n_vars = len(vars_to_plot)
        print(f"📊 Analyzing {n_vars} variables: {vars_to_plot}")

        if n_vars == 0:
            print("❌ Error: No numeric variables found to plot.")
        else:
            # Dynamic Grid Calculation (3 Columns wide)
            n_cols = 3
            n_rows = math.ceil(n_vars / n_cols)

            # Generate a distinct color for each variable
            colors = sns.color_palette("husl", n_vars)

            # Helper function to create the figure grid
            def setup_grid(title):
                fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 5, n_rows * 4))
                axes = axes.flatten() # Flatten 2D grid to 1D list

                # Hide empty subplots if vars < rows*cols
                for i in range(n_vars, len(axes)):
                    fig.delaxes(axes[i])

                plt.suptitle(title, fontsize=22, fontweight='bold', y=1.01)
                return fig, axes

            # ==========================================
            # FIGURE 1: COMBINED (Violin + Swarm + Box)
            # ==========================================
            sns.set_theme(style="whitegrid", context="talk", rc={"font.family": "serif"})
            fig, axes = setup_grid("Combined Analysis: Violin & Swarm Plots (All Variables)")

            for i, col in enumerate(vars_to_plot):
                ax = axes[i]

                # 1. Violin (Shape)
                sns.violinplot(y=df[col], ax=ax, inner=None, color=colors[i], alpha=0.3, linewidth=0)

                # 2. Swarm (Data Points)
                # Adjust point size based on dataset size to avoid overcrowding
                pt_size = 3 if len(df) > 100 else 5
                sns.swarmplot(y=df[col], ax=ax, color="black", size=pt_size, alpha=0.6)

                # 3. Box (Statistics)
                sns.boxplot(y=df[col], ax=ax, width=0.15,
                            boxprops={'facecolor':'none', 'edgecolor':'black', 'zorder': 2},
                            whiskerprops={'color':'black', 'linewidth': 1.5},
                            capprops={'color':'black', 'linewidth': 1.5},
                            medianprops={'color':'white', 'linewidth': 2},
                            showfliers=False)

                ax.set_ylabel(col, fontsize=12, fontweight='bold')
                ax.set_title(col, fontsize=14, pad=10)
                sns.despine(ax=ax, left=True)

            plt.tight_layout()
            plt.show()

            # ==========================================
            # FIGURE 2: VIOLIN PLOTS ONLY
            # ==========================================
            fig, axes = setup_grid("Distribution Analysis: Violin Plots Only (All Variables)")

            for i, col in enumerate(vars_to_plot):
                ax = axes[i]
                # inner='quartile' draws lines for 25%, 50%, 75% inside the violin
                sns.violinplot(y=df[col], ax=ax, inner='quartile', color=colors[i], alpha=0.8)

                ax.set_ylabel(col, fontsize=12, fontweight='bold')
                ax.set_title(col, fontsize=14)
                sns.despine(ax=ax, left=True)

            plt.tight_layout()
            plt.show()

            # ==========================================
            # FIGURE 3: SWARM PLOTS ONLY
            # ==========================================
            fig, axes = setup_grid("Data Density Analysis: Swarm Plots Only (All Variables)")

            for i, col in enumerate(vars_to_plot):
                ax = axes[i]
                pt_size = 4 if len(df) > 100 else 6
                sns.swarmplot(y=df[col], ax=ax, color=colors[i], size=pt_size, alpha=0.8)

                ax.set_ylabel(col, fontsize=12, fontweight='bold')
                ax.set_title(col, fontsize=14)
                sns.despine(ax=ax, left=True)
                ax.grid(True, axis='y', linestyle='--', alpha=0.3)

            plt.tight_layout()
            plt.show()

    except Exception as e:
        print(f"❌ Error processing file: {e}")
```
